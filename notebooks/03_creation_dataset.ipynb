```python
import pandas as pd
from pathlib import Path

# Chemins relatifs sécurisés
DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# 1. Chargement de la base préparée
df_complet = pd.read_parquet(PROCESSED_DIR / "dataset_train_fusion.parquet")

# 2. Séparation : on retire la variable cible TARGET
X = df_complet.drop(columns=['TARGET'])

# 3. Positionnement de l'ID client en INDEX (crucial pour le dashboard et la conformité des features)
if 'SK_ID_CURR' in X.columns:
    X = X.set_index('SK_ID_CURR')

# 4. Échantillonnage de 50 clients représentatifs
X_sample = X.sample(n=50, random_state=42)

# 5. Exportation au format Parquet avec l'index conservé
# On utilise processed/ pour être aligné avec le chemin lu dans dashboard.py et serving_mlflow.py
output_path = PROCESSED_DIR / "X_test_sample.parquet"
X_sample.to_parquet(output_path, index=True)

print(f"✅ Échantillon sauvegardé avec succès sous : {output_path}")
print(f"Dimensions : {X_sample.shape[0]} lignes, {X_sample.shape[1]} features (index = SK_ID_CURR)")
```
